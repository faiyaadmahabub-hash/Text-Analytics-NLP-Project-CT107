*<small>Step 1: Loading Data_1.txt</small>*

```python
# Loading Data_1.txt

with open("../Data/Data_1.txt", "r", encoding="utf-8") as file:
    text = file.read()

print(text)
```

*<small>Step 2: Tokenization</small>*

```python
# import re to to search for pattern in word
import re

# import nltk to work with language & text data
import nltk

# import word_tokenize which splits the text into words & punctuation
from nltk.tokenize import word_tokenize

# download punkt dataset - required to split the text into words as it containes rule for punctuation & sentence boundaries
nltk.download('punkt')
nltk.download('punkt_tab')

# tokenize the text
nltk_tokens = word_tokenize(text)

# print tokens (checking purposes)
print(nltk_tokens)
```

*<small>Step 3: Remove Stopword</small>*

```python

# import the list of stopwords from nltk
from nltk.corpus import stopwords

# import string module for punctuation removal later on
import string

# download nltk's stopwords dataset
nltk.download('stopwords')

# create a set of english words for faster lookup
stop_words = set(stopwords.words('english'))

# Convert all tokens to lowercase
tokens_lower = [word.lower() for word in nltk_tokens]

# Remove all tokens that are in the stopwords list
filtered_tokens = [word for word in tokens_lower if word not in stop_words]

# check if stopwords were actually found in the text
print("Stop Words Found in Corpus:\n")
found_stopwords = [word for word in tokens_lower if word in stop_words]
print(set(found_stopwords))

# print the tokens after removing stopwords
print("\nTokens after Stop Word Removal:\n")
print(filtered_tokens)
```

*<small>Step 4: Remove Punctuation</small>*

```python
# removing the punctuation
final_tokens = [word for word in filtered_tokens if word not in string.punctuation]

# print the tokens after removing punctuation
print("Final Tokens after Removing Stop Words and Punctuation:\n")
print(final_tokens)
```
