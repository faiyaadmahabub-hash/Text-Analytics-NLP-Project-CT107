*<small>Q2. Form Word Stemming</small>*

*<small>Step 1: Loading the Dataset</small>*

```python
# Loading Data_1.txt

with open("../Data/Data_1.txt", "r", encoding="utf-8") as file:
    text = file.read()

print(text)
```

*<small>Step 2: Tokenization Using NLTK</small>*

```python
# import NLTK to work with language & text data
import nltk

# download punkt dataset - required to split the text into words as it contains
# rules
# for punctuation & sentence boundaries
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
# import word_tokenize which splits the text into words & punctuation
from nltk.tokenize import word_tokenize

# tokenize the text
nltk_tokens = word_tokenize(text)

# print nltk tokens
print("Tokens using NLTK: \n")
print(nltk_tokens)
```

*<small>Step 3: Removing Stop Words & Punctuation</small>*

```python
# import nltk 
# import string - provides useful text constant like punctuation
import nltk, string

# download stopwords dataset
nltk.download('stopwords')

# import stopwords - to remove common & unimportant words
from nltk.corpus import stopwords

# convert nltk_tokens to lowercase
tokens_lower = [word.lower() for word in nltk_tokens]

# combine stopwords + punctuation 
stopTokens = stopwords.words("english") + list(string.punctuation)

# prepares a list to store cleaned words
filteredTokens = []

# remove stopwords + punctuation
for w in tokens_lower:
    if w not in stopTokens:
        filteredTokens.append(w)

# outputs
# prints tokens
print("Tokens: \n", nltk_tokens)

# prints no. of tokens
print("\nTotal number of words in this text corpus: ", len(nltk_tokens))

# prints no. of filtered tokens
print("\nThere are", len(filteredTokens), "words after removing stop words and punctuation\n")

# prints filtered tokens
print("Filtered Tokens: \n", filteredTokens)

# prints stopwords
print("\nStop Tokens (Stopwords + Punctuation): \n", stopTokens)
```

*<small>Step 4: Demonstrate Word Stemming Using Regular Expression</small>*

```python
# import regular expression module
import re

# define the regular expression stemmer function
def regex_stem(word):
    pattern = r'^(.*?)(ing|ly|ed|ious|ies|ive|es|s|ment)?$'
    stem, suffix = re.findall(pattern, word)[0]
    return stem

# apply stemming on filtered tokens
RegexStemmedTokens = [regex_stem(word) for word in filteredTokens]

# outputs
print("Filtered Tokens:\n", filteredTokens)
print("\nStemmed Tokens:\n", RegexStemmedTokens)
```

*<small>Step 5: Demonstrate Word Stemming Using Porter Stemmer</small>*

```python
# imports porter stemmer from nltk - it's a stemming algorithm that removes 
# common endings
from nltk.stem import PorterStemmer

# initialize porter stemmer - creating a porter stemmer object
ps = PorterStemmer()

# apply stemming on filtered tokens
porterStemmedTokens = [ps.stem(word) for word in filteredTokens]

# outputs
print("Filtered Tokens:\n", filteredTokens)
print("\nPorter Stemmed Tokens:\n", porterStemmedTokens)
```

*<small>Step 6: Demonstrate Word Stemming Using Lancaster Stemmer</small>*

```python
# imports lancaster stemmer from nltk
from nltk.stem import LancasterStemmer

# initialize lancaster stemmer - creating a lancaster stemmer object
ls = LancasterStemmer()

# apply stemming on filtered tokens
lancasterStemmedTokens = [ls.stem(word) for word in filteredTokens]

# outputs
print("Filtered Tokens:\n", filteredTokens)
print("\nLancaster Stemmed Tokens:\n", lancasterStemmedTokens)
```

*<small>Q5. Tokenization Using An Alternative Approach</small>*  
*<small>Step 1: TextBlob Tokenization Approach</small>*

```python
# import textblob class from textblob library
from textblob import TextBlob

# make a textblob object using the text
blob = TextBlob(text)

# tokenize the text into individual words
textblob_tokens = blob.words

# display the generated tokens
print("Tokens using TextBlob:\n")
print(textblob_tokens)
```

*<small>Step 2: Evaluation of TextBlob Approach</small>*

```python
# import the string module to access predefined constants
import string

# computes the total number of tokens
total = len(textblob_tokens)

# counts the tokens with punctuation attached (e.g., "word.", "text,")
punct_attached = sum(
    1 for t in textblob_tokens
    if any(p in t for p in string.punctuation) and not all(c in string.punctuation for c in t)
)

# counts the unique tokens (vocabulary size)
unique = len(set(textblob_tokens))

# counts the hyphenated words in the original text that were split incorrectly
hyphen_words = [word for word in text.split() if '-' in word]
hyphen_split_count = sum(1 for hw in hyphen_words if hw not in textblob_tokens)

# print results
print("--- TextBlob Tokenizer ---")
print("Total tokens:", total)
print("Tokens with punctuation attached:", punct_attached)
print("Hyphenated words split incorrectly:", hyphen_split_count)
print("Unique tokens:", unique)
```
