*<small>Q2. Form Word Stemming<small>*

*<small>Step 1: Loading Dataset<small>*

```python
# Loading Data_1.txt

with open("../Data/Data_1.txt", "r", encoding="utf-8") as file:
    text = file.read()

print(text)
```

*<small>Step 2: Tokenization<small>*

```python
# import re to to search for pattern in word -
import re

# import nltk to work with language & text data -
import nltk

# import word_tokenize which splits the text into words & punctuation 
from nltk.tokenize import word_tokenize

# download punkt dataset - required to split the text into words as it containes rule for punctuation & sentence boundaries -
nltk.download('punkt')
nltk.download('punkt_tab')

# tokenize the text -
tokens = word_tokenize(text)

# print tokens (checking purposes) -
print(tokens)
```

*<small>Step 3: Demonstrate Word Stemming Using Regular Expression<small>*

```python
# define the regular expression stemmer function
def regex_stem(word):
    pattern = r'^(.*?)(ing|ly|ed|ious|ies|ive|es|s|ment)?$'
    stem, suffix = re.findall(pattern, word)[0]
    return stem

# print stemmer results
print([regex_stem(word) for word in tokens])
```

*<small>Step 4: Demonstrate Word Stemming Using Porter Stemmer<small>*

```python
# imports port stemmer from nltk - it's a stemming algorithm the removes common endings
from nltk.stem import PorterStemmer

# initialize porter stemmer - creating a port stemmer object
ps = PorterStemmer()

# this will stem each word and print the result
print([ps.stem(word) for word in tokens])
```

*<small>Step 5: Demonstrate Word Stemming Using Lancaster Stemmer<small>*

```python
# imports lacaster stemmer from nltk
from nltk.stem import LancasterStemmer

# initialize lancaster stemmer - creating a lancaster stemmer object
ls = LancasterStemmer()

# this will stem each word and print the result
print([ls.stem(word) for word in tokens])
```

*<small>Q5. Tokenization Using An Alternative Approach: TextBlob<small>*

*<small>Step 1: TextBlob Tokenization Approach<small>*

```python
# import textblob class from textblob library
from textblob import TextBlob

# make a textblob object using the text
blob = TextBlob(text)

# tokenize the text into individual words
textblob_tokens = blob.words

# display the generated tokens
print("Tokens using TextBlob:\n")
print(textblob_tokens)
```

*<small>Step 2: Evaluation of TextBlob Approach<small>*

```python
# import the string module to access predefined constants
import string

# computes the total number of tokens
total = len(textblob_tokens)

# counts the tokens with punctuation attached (e.g., "word.", "text,")
punct_attached = sum(
    1 for t in textblob_tokens
    if any(p in t for p in string.punctuation) and not all(c in string.punctuation for c in t)
)

# counts the unique tokens (vocabulary size)
unique = len(set(textblob_tokens))

# counts the hyphenated words in the original text that were split incorrectly
hyphen_words = [word for word in text.split() if '-' in word]
hyphen_split_count = sum(1 for hw in hyphen_words if hw not in textblob_tokens)

# print results
print("--- TextBlob Tokenizer ---")
print("Total tokens:", total)
print("Tokens with punctuation attached:", punct_attached)
print("Hyphenated words split incorrectly:", hyphen_split_count)
print("Unique tokens:", unique)
```
